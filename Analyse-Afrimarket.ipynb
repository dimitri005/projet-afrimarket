```python
# ================================================
# ANALYSE DE STRUCTURE DU DATASET
# ================================================
# Cette cellule affiche la forme (dimensions) du dataset :
# - Nombre de lignes (observations/commandes)
# - Nombre de colonnes (variables/caractéristiques)
# 'shape' est un attribut pandas qui retourne (nombre_lignes, nombre_colonnes)

print(f"\nSTRUCTURE DES DONNÉES : {afrimarket.shape[0]} lignes et {afrimarket.shape[1]} colonnes.")
```

```python
# ================================================
# VÉRIFICATION DES TYPES DE DONNÉES
# ================================================
# 'dtypes' affiche le type de donnée de chaque colonne :
# - object = texte (string)
# - int64 = nombre entier (integer)
# - float64 = nombre décimal (decimal)
# Cette vérification est importante pour s'assurer que les données
# sont dans le bon format avant toute analyse

print("\nTYPES DES DONNÉES : Vérification des formats colonnes\n")
print(afrimarket.dtypes)
```

```python
# ================================================
# VÉRIFICATION DES VALEURS MANQUANTES
# ================================================
# 'isnull()' détecte les cellules vides (NaN = Not a Number)
# 'sum()' compte le nombre de valeurs manquantes par colonne
# Une valeur manquante est une donnée qui n'a pas été enregistrée
# Il est crucial de les identifier car elles peuvent fausser les analyses

print("VALEURS MANQUANTES :")
val_mqte = afrimarket.isnull().sum()  # Compte les valeurs nulles par colonne
print(val_mqte[val_mqte > 0] if val_mqte.any() else "Aucune donnée manquante.")
```

```python
# ================================================
# DÉTECTION DES VALEURS ABÉRANTES (ANOMALIES)
# ================================================
# Les valeurs aberrantes sont des données impossibles ou incohérentes
# qui proviennent d'erreurs de saisie ou de dysfonctionnements.
# On détecte ici 3 types d'anomalies financières :

# 1. prix_unitaire < 0 : Un prix ne peut pas être négatif
prix_neg = (afrimarket['prix_unitaire'] < 0).sum()

# 2. quantite <= 0 : Une quantité commandée doit être au moins 1
quantite_neg = (afrimarket['quantite'] <= 0).sum()

# 3. remise < 0 ou remise > 1 : Une remise doit être entre 0% et 100%
remise_neg = ((afrimarket['remise'] < 0) | (afrimarket['remise'] > 1)).sum()

print("\nVALEURS ABÉRANTES :")
print(f"Prix unitaire négatif : {prix_neg}")
print(f"Quantité négative ou nulle : {quantite_neg}")
print(f"Remise négative ou supérieure à 1 : {remise_neg}")
```

<center><p><font size="5" color='cyan' class="alert alert-primary"><b># 2. PHASE DE NETTOYAGE (DATA CLEANING)</b></font></p><center>

```python
# ================================================
# PHASE DE NETTOYAGE DES DONNÉES (DATA CLEANING)
# ================================================
# Cette étape transforme le dataset brut en données fiables et exploitables
# Chaque opération corrige un problème identifié lors de l'audit

# -----------------------------------------------
# 1. SUPPRESSION DES DOUBLONS
# -----------------------------------------------
# On supprime les lignes en double basées sur l'ID de commande
# Chaque commande doit apparaître une seule fois
afrimarket = afrimarket.drop_duplicates(subset='id_commande')

# -----------------------------------------------
# 2. CONVERSION DES DATES AU FORMAT DATETIME
# -----------------------------------------------
# 'pd.to_datetime()' convertit le texte de la colonne 'date_commande'
# en format date/heure pour permettre des calculs temporels
afrimarket['date_commande'] = pd.to_datetime(afrimarket['date_commande'])

# -----------------------------------------------
# 3. UNIFORMISATION DES CATÉGORIES
# -----------------------------------------------
# On standardise les catégories pour éviter les doublons :
# - 'strip()' : supprime les espaces au début/fin
# - 'str.lower()' : met tout en minuscules
# - 'str.capitalize()' : met la première lettre en majuscule
# - 'replace()' : corrige les fautes d'orthographe spécifiques
afrimarket['categorie'] = afrimarket['categorie'].str.strip().str.lower().str.capitalize()
afrimarket['categorie'] = afrimarket['categorie'].str.replace('Electronique', 'Électronique')

# -----------------------------------------------
# 4. CORRECTION DES NOMS DE VILLES
# -----------------------------------------------
# On crée un dictionnaire de corrections pour uniformes les noms de villes
# Exemple : "Kinshassa" -> "Kinshasa", "Abidjan " -> "Abidjan"
corrections_villes = {
    'Kinshassa': 'Kinshasa',
    'Abidjan ': 'Abidjan',
    'Dakar_': 'Dakar'
}
# 'replace()' applique les corrections, puis 'strip()' et 'capitalize()' finalisent
afrimarket['ville'] = afrimarket['ville'].replace(corrections_villes).str.strip().str.capitalize()

# -----------------------------------------------
# 5. SUPPRESSION DES VALEURS ABÉRANTES
# -----------------------------------------------
# On filtre pour ne garder que les lignes avec des données cohérentes :
# - prix_unitaire > 0 : prix strictement positif
# - quantite > 0 : quantité strictement positive
afrimarket = afrimarket[(afrimarket['prix_unitaire'] > 0) & (afrimarket['quantite'] > 0)]

# -----------------------------------------------
# 6. CORRECTION DES REMISES
# -----------------------------------------------
# Les remises doivent être entre 0 et 1 (0% à 100%)
# 'abs()' convertit les valeurs négatives en positives
# 'clip(upper=1)' limite les valeurs à un maximum de 1
afrimarket['remise'] = afrimarket['remise'].abs()
afrimarket['remise'] = afrimarket['remise'].clip(upper=1)

# -----------------------------------------------
# 7. STANDARDISATION DES STATUTS DE COMMANDES
# -----------------------------------------------
# On met les statuts au même format : première lettre majuscule, pas d'espaces
# Exemple : "livrée" -> "Livrée", "ANNULÉE" -> "Annulée"
afrimarket['statut_commande'] = afrimarket['statut_commande'].str.strip().str.capitalize()

# -----------------------------------------------
# 8. VÉRIFICATION FINALE APRÈS NETTOYAGE
# -----------------------------------------------
# On affiche un résumé pour confirmer que le nettoyage a réussi
print("\n=== VÉRIFICATION FINALE APRÈS NETTOYAGE ===")
print(f"Structure des données : {afrimarket.shape[0]} lignes et {afrimarket.shape[1]} colonnes.")
print("\nTypes des données :")
print(f"{afrimarket.dtypes}\n")

# Affichage du dataset nettoyé
afrimarket
```

<center><p><font size="5" color='cyan' class="alert alert-primary"><b># 3. PHASE DE FEATURE ENGINEERING</b></font></p><center>

```python
# ================================================
# PHASE DE FEATURE ENGINEERING (CRÉATION DE KPIs)
# ================================================
# Cette étape crée de nouvelles colonnes calculées pour enrichir l'analyse
# Ces indicateurs permettent de mesurer la performance et la rentabilité

# -----------------------------------------------
# 1. CHIFFRE D'AFFAIRES NET (CA_Net)
# -----------------------------------------------
# Formule : (Prix unitaire × Quantité) × (1 - Remise)
# Cela représente le revenu RÉEL après application des remises
# Exemple : Un produit à 100€ avec 10% de remise = 100 × 1 × 0.90 = 90€
Afri_Net['CA_Net'] = (Afri_Net['prix_unitaire'] * Afri_Net['quantite']) * (1 - Afri_Net['remise'])

# -----------------------------------------------
# 2. MARGE BRUTE (Marge_Brute)
# -----------------------------------------------
# La marge brute représente le bénéfice après avoir soustrait le coût des marchandises
# On suppose que le coût des marchandises = 60% du prix unitaire
# Formule : CA_Net - (Quantité × Prix × 0.60)
Afri_Net['Marge_Brute'] = Afri_Net['CA_Net'] - (Afri_Net['quantite'] * Afri_Net['prix_unitaire'] * 0.6)

# -----------------------------------------------
# 3. PROFIT NET (profit_net)
# -----------------------------------------------
# Le profit net = Marge brute - (Coût marketing + Coût livraison)
# C'est le bénéfice RÉEL après tous les coûts opérationnels
Afri_Net['profit_net'] = Afri_Net['Marge_Brute'] - (Afri_Net['cout_marketing'] + Afri_Net['cout_livraison'])

# -----------------------------------------------
# 4. EXTRACTION DU MOIS (mois)
# -----------------------------------------------
# On extrait le nom du mois de la date de commande pour l'analyse temporelle
# 'dt.month_name()' retourne le nom complet du mois (ex: "Décembre")
Afri_Net['mois'] = Afri_Net['date_commande'].dt.month_name()

# -----------------------------------------------
# 5. INDICATEUR DE RETOUR (indicateur_retour)
# -----------------------------------------------
# Flag binaire : 1 si la commande est Annulée ou Retournée, 0 sinon
# Permet de calculer facilement le taux de retour/annulation
Afri_Net['indicateur_retour'] = Afri_Net['statut_commande'].apply(
    lambda x: 1 if x in ['Annulée', 'Retournée'] else 0
)

# -----------------------------------------------
# 6. NOMBRE DE COMMANDES PAR CLIENT
# -----------------------------------------------
# On compte combien de fois chaque client a passé commande
# 'groupby()' groupe par client, puis 'transform("count")' compte
Afri_Net['nombre_commandes_par_client'] = Afri_Net.groupby('id_client')['id_commande'].transform('count')

# -----------------------------------------------
# 7. VALEUR VIE CLIENT (CLV - Customer Lifetime Value)
# -----------------------------------------------
# La CLV représente le profit total généré par un client sur toute sa période d'achat
# On fait la somme des profits nets par client
Afri_Net['valeur_vie_client'] = Afri_Net.groupby('id_client')['profit_net'].transform('sum')

# Affichage du résultat
Afri_Net.head()
```

```python
# ================================================
# ANALYSE PAR CATÉGORIE (CA, Marge, Taux de Retour)
# ================================================
# Cette cellule group les données par catégorie de produit pour voir
# quelle catégorie génère le plus de chiffre d'affaires et de marge

# 'groupby()' regroup les lignes par catégorie
# 'agg()' applique des fonctions d'agrégation sur chaque colonne
cat_analysis = Afri_Net.groupby('categorie').agg({
    'CA_Net': 'sum',                  # Somme du CA par catégorie
    'Marge_Brute': 'sum',             # Somme de la marge par catégorie
    'indicateur_retour': 'mean'        # Moyenne du flag retour = taux de retour
}).reset_index()  # reset_index() convertit le résultat en DataFrame classique

# Conversion du taux de retour en pourcentage (0.10 -> 10%)
cat_analysis['taux_retour_cat'] = cat_analysis['indicateur_retour'] * 100

# Tri par CA décroissant (plus gros CA en premier)
cat_analysis = cat_analysis.sort_values(by='CA_Net', ascending=False)
cat_analysis
```

```python
# ================================================
# ANALYSE GÉOGRAPHIQUE (PAR VILLE)
# ================================================
# Cette cellule analyse la performance par ville pour identifier
# quelles zones géographiques génèrent le plus de CA et de profit

# Groupement par ville avec agrégation de plusieurs métriques
ville_analysis = Afri_Net.groupby('ville').agg({
    'CA_Net': 'sum',                 # CA total par ville
    'profit_net': 'sum',             # Profit total par ville
    'indicateur_retour': 'mean'       # Taux d'annulation/retour par ville
}).reset_index()

# Conversion du taux en pourcentage pour lisibilité
ville_analysis['taux_annulation_ville'] = ville_analysis['indicateur_retour'] * 100

# Tri par CA décroissant (ville la plus rentable en premier)
ville_analysis = ville_analysis.sort_values(by='CA_Net', ascending=False)
ville_analysis
```

<center><p><font size="5" color='cyan' class="alert alert-primary"># 7. PHASE D'ANALYSE DE LA PERFORMANCE MARKETING</font></p><center>

```python
# ================================================
# ANALYSE DE LA PERFORMANCE MARKETING (ROI & RÉTENTION)
# ================================================
# Cette cellule évalue l'efficacité des différents canaux marketing
# en calculant le ROI et le taux de rétention pour chaque canal

# -----------------------------------------------
# 1. AGRÉGATION DES DONNÉES PAR CANAL
# -----------------------------------------------
# Groupement par canal marketing avec plusieurs métriques
mkt_perf = Afri_Net.groupby('canal_marketing').agg({
    'CA_Net': 'sum',                 # CA généré par le canal
    'cout_marketing': 'sum',         # Coût dépensé sur ce canal
    'id_client': 'nunique'           # Nombre de clients uniques Acquisition
}).reset_index()

# -----------------------------------------------
# 2. CALCUL DU ROI (Return on Investment)
# -----------------------------------------------
# Formule : (CA - Coût) / Coût
# Un ROI de 10 signifie que pour 1€ dépensé, on génère 10€ de CA
# ROI > 0 = rentable, ROI < 0 = déficitaire
mkt_perf['ROI'] = (mkt_perf['CA_Net'] - mkt_perf['cout_marketing']) / mkt_perf['cout_marketing']

# -----------------------------------------------
# 3. CALCUL DU TAUX DE RÉTENTION PAR CANAL
# -----------------------------------------------
# Un client fidèle = un client avec plus d'une commande
# On compte ces clients par canal
clients_fidele_par_canal = Afri_Net[Afri_Net['nombre_commandes_par_client'] > 1].groupby('canal_marketing')['id_client'].nunique()

# Ratio = clients fidèles / total clients du canal × 100
mkt_perf['taux_retention'] = (mkt_perf['canal_marketing'].map(clients_fidele_par_canal) / mkt_perf['id_client']) * 100

# -----------------------------------------------
# 4. AFFICHAGE DU TABLEAU FINAL
# -----------------------------------------------
# Sélection et tri des colonnes pertinentes
mkt_perf_final = mkt_perf[['canal_marketing', 'CA_Net', 'cout_marketing', 'ROI', 'taux_retention']]
print("Tableau de Performance Marketing :")
display(mkt_perf_final.sort_values('ROI', ascending=False).style.format({
    'CA_Net': '{:,.0f} $',           # Formatage : séparateur de milliers + symbole $
    'cout_marketing': '{:,.0f} $',
    'ROI': '{:.2f}',                 # 2 décimales
    'taux_retention': '{:.1f}%'      # 1 décimale + symbole %
}))
```

<center><p><font size="5" color='cyan' class="alert alert-primary">#8. PHASE D'ANALYSE CLIENTS & SEGMENTATION</font></p><center>

```python
# ================================================
# ANALYSE CLIENTS & SEGMENTATION (LOI DE PARETO)
# ================================================
# Cette cellule analyse la base client pour identifier les segments clés
# et appliquer la loi de Pareto (80/20)

# -----------------------------------------------
# 1. NOMBRE TOTAL DE CLIENTS UNIQUES
# -----------------------------------------------
# 'nunique()' compte le nombre de valeurs distinctes (sans doublons)
total_clients = Afri_Net['id_client'].nunique()

# -----------------------------------------------
# 2. CALCUL DU TAUX DE CLIENTS RÉCURRENTS
# -----------------------------------------------
# Un client récurrent = un client avec plus d'une commande
# On compte combien de clients ont commandé plusieurs fois
clients_commandes = Afri_Net.groupby('id_client')['id_commande'].nunique()
nb_clients_recurrents = (clients_commandes > 1).sum()
pct_recurrents = (nb_clients_recurrents / total_clients) * 100

# -----------------------------------------------
# 3. TOP 10 CLIENTS (VIP)
# -----------------------------------------------
# Identification des 10 clients qui génèrent le plus de CA
# 'sort_values(ascending=False)' trie du plus grand au plus petit
top_10_clients = Afri_Net.groupby('id_client')['CA_Net'].sum().sort_values(ascending=False).head(10).reset_index()

# -----------------------------------------------
# 4. ANALYSE DE PARETO (80/20)
# -----------------------------------------------
# La loi de Pareto dit que 80% du CA vient de 20% des clients
# On calcule ici le pourcentage cumulé du CA et des clients
# Trie des clients par CA décroissant
df_pareto = Afri_Net.groupby('id_client')['CA_Net'].sum().sort_values(ascending=False).reset_index()

# Ajout d'une colonne avec le CA cumulé
df_pareto['cum_ca'] = df_pareto['CA_Net'].cumsum()

# Pourcentage cumulé du CA (par rapport au total)
df_pareto['cum_ca_pct'] = 100 * df_pareto['cum_ca'] / df_pareto['CA_Net'].sum()

# Pourcentage cumulé des clients (1er client = 1/N, 2ème = 2/N...)
df_pareto['cum_clients_pct'] = 100 * (df_pareto.index + 1) / len(df_pareto)

# -----------------------------------------------
# AFFICHAGE DES RÉSULTATS
# -----------------------------------------------
print(f"Nombre total de clients : {total_clients}\n")
print(f"Taux de clients récurrents : {pct_recurrents:.1f}%")
print("\nTop 10 Clients (VIP) :")
top_10_clients
```
